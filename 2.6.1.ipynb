```python
import numpy as np
import matplotlib.pyplot as plt

# Функция
def f(x):
    return np.log10(x) - np.cos(x)

# Метод половинного деления
def bisection(f, a, b, tol=1e-4, max_iter=100):
    if f(a) * f(b) >= 0:
        raise ValueError("f(a) и f(b) должны иметь разные знаки")
    
    print("-" * 70)
    print(f"{'n':>3} | {'a':>12} | {'b':>12} | {'c':>12} | {'f(c)':>15}")
    print("-" * 70)
    
    for i in range(max_iter):
        c = (a + b) / 2
        fc = f(c)
        print(f"{i+1:>3} | {a:>12.6f} | {b:>12.6f} | {c:>12.6f} | {fc:>15.2e}")
        
        if abs(fc) < tol or (b - a) / 2 < tol:
            break
        elif f(a) * fc < 0:
            b = c
        else:
            a = c
    
    print("-" * 70)
    return c

# Вычисление
print("=" * 70)
print("Упражнение 1. Метод половинного деления")
print("Уравнение: lg x - cos x = 0")
print("=" * 70)

root = bisection(f, 1, 1.5, tol=1e-4)

print(f"\nКорень с точностью 10^-4: x = {root:.6f} → {root:.4f}")
print(f"Проверка: f(x) = {f(root):.2e}")

# График
x_plot = np.linspace(0.5, 2, 500)
y_plot = f(x_plot)

plt.figure(figsize=(10, 5))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='f(x) = lg x - cos x')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5)
plt.axvline(x=root, color='r', linestyle='--', label=f'Корень x = {root:.4f}')
plt.plot(root, f(root), 'ro', markersize=8)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Графическое отделение корня уравнения lg x - cos x = 0')
plt.legend()
plt.show()
```
