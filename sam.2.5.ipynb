```python
import numpy as np
from scipy.optimize import fsolve, root_scalar

def F(x):
    return 2*x**2 - 2*x - 5

def F1(x):
    return 4*x - 2

def phi(x):
    return np.sqrt(x + 2.5)

def simple_iteration(x0, eps=1e-6, max_iter=100):
    history = [x0]
    for n in range(max_iter):
        x1 = phi(x0)
        history.append(x1)
        if abs(x1 - x0) < eps:
            return x1, n + 1, history
        x0 = x1
    return x0, max_iter, history

def combined_method(a, b, eps=1e-6, max_iter=100):
    if F(a) * 4 > 0: 
        c = a
        x_tan = a
        x_chord = b
    else:
        c = b
        x_tan = b
        x_chord = a
    
    for n in range(max_iter):
        x_tan_new = x_tan - F(x_tan) / F1(x_tan)
        x_chord_new = (c * F(x_chord) - x_chord * F(c)) / (F(x_chord) - F(c))
        
        if abs(x_tan_new - x_chord_new) / 2 < eps:
            return (x_tan_new + x_chord_new) / 2, n + 1
        x_tan, x_chord = x_tan_new, x_chord_new
    return (x_tan + x_chord) / 2, max_iter


root_simple, n_simple, _ = simple_iteration(2.0, eps=1e-6)


root_combined, n_combined = combined_method(1.0, 3.0, eps=1e-6)


root_fsolve = fsolve(F, 2)[0]
result_brent = root_scalar(F, bracket=[1, 3], method='brentq')
root_brent = result_brent.root
iter_brent = result_brent.iterations


sqrt11 = np.sqrt(11)
root_analytical = (1 + sqrt11) / 2

print("=" * 60)
print("Задание 5. Инструментальные пакеты (SciPy)")
print("=" * 60)

print(f"\n1. fsolve (начальное приближение x0 = 2):")
print(f"   Корень: {root_fsolve:.10f}")
print(f"   F(корень): {F(root_fsolve):.2e}")

print(f"\n2. root_scalar (метод Brent, отрезок [1, 3]):")
print(f"   Корень: {root_brent:.10f}")
print(f"   F(корень): {F(root_brent):.2e}")
print(f"   Сходимость: {iter_brent} итераций")

print(f"\nАналитическое решение (для сравнения):")
print(f"   x = (1 + √11)/2 = {root_analytical:.10f}")

print("\n" + "=" * 60)
print("Сравнение результатов всех методов:")
print("=" * 60)
print(f"Метод половинного деления:    2.158200")
print(f"Метод простой итерации:       {root_simple:.10f}")
print(f"Комбинированный метод:        {root_combined:.10f}")
print(f"Инструментальный (fsolve):    {root_fsolve:.10f}")
print(f"Инструментальный (Brent):     {root_brent:.10f}")
print(f"Аналитически:                 {root_analytical:.10f}")
```
