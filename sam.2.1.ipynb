```python
import numpy as np
import matplotlib.pyplot as plt

def F(x):
    """Функция F(x) = 2x² - 2x - 5"""
    return 2*x**2 - 2*x - 5

print("=" * 60)
print("Задание 1. Отделение корней")
print("=" * 60)

print("\nТаблица значений F(x):")
print("   x   |   F(x)   |  Знак")
print("-------|----------|-------")

prev_x = -5.0
prev_F = F(prev_x)
roots_segments = []

for x in np.arange(-4.5, 5.1, 0.5):
    current_F = F(x)
    sign = "+" if current_F > 0 else ("-" if current_F < 0 else "0")
    print(f" {x:4.1f} | {current_F:8.3f} |   {sign}")
    
    if prev_F * current_F < 0:
        roots_segments.append((prev_x, x))
        print(f"\n*** Корень на отрезке [{prev_x}; {x}] ***\n")
    
    prev_x, prev_F = x, current_F

print("\nНайденные отрезки с корнями:")
for a, b in roots_segments:
    print(f"  [{a}; {b}]")

x_plot = np.linspace(-3, 4, 500)
y_plot = F(x_plot)

plt.figure(figsize=(12, 6))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='F(x) = 2x² - 2x - 5')
plt.axhline(y=0, color='k', linestyle='--', linewidth=1)
plt.axvline(x=0, color='k', linestyle='--', linewidth=0.5)
plt.grid(True, alpha=0.3)

# Выделяем отрезки с корнями
for a, b in roots_segments:
    plt.axvspan(a, b, alpha=0.2, color='green')

plt.xlabel('x')
plt.ylabel('F(x)')
plt.title('Графическое отделение корней уравнения 2x² - 2x - 5 = 0')
plt.legend()
plt.show()
```
