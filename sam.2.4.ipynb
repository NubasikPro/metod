```python
def F(x):
    return 2*x**2 - 2*x - 5

def F1(x):
    """Первая производная F'(x) = 4x - 2"""
    return 4*x - 2

def F2(x):
    """Вторая производная F''(x) = 4"""
    return 4

def combined_method(a, b, eps=1e-6, max_iter=100):

    if F(a) * F2(a) > 0:
        c = a
        x_tan = a     
        x_chord = b 
        print(f"\nF({a})·F''({a}) > 0 → c = a = {a}")
        print(f"Начальное приближение для касательных: x_tan0 = {a}")
        print(f"Начальное приближение для хорд: x_chord0 = {b}")
    else:
        c = b
        x_tan = b     
        x_chord = a    
        print(f"\nF({b})·F''({b}) > 0 → c = b = {b}")
        print(f"Начальное приближение для касательных: x_tan0 = {b}")
        print(f"Начальное приближение для хорд: x_chord0 = {a}")
    
    history_tan = [x_tan]
    history_chord = [x_chord]
    errors = []
    
    for n in range(max_iter):

        x_tan_new = x_tan - F(x_tan) / F1(x_tan)
        
  
        x_chord_new = (c * F(x_chord) - x_chord * F(c)) / (F(x_chord) - F(c))
        
        history_tan.append(x_tan_new)
        history_chord.append(x_chord_new)
        
 
        error = abs(x_tan_new - x_chord_new) / 2
        errors.append(error)
        
        if error < eps:
            root = (x_tan_new + x_chord_new) / 2
            return root, n + 1, history_tan, history_chord, errors
        
        x_tan, x_chord = x_tan_new, x_chord_new
    
    root = (history_tan[-1] + history_chord[-1]) / 2
    return root, max_iter, history_tan, history_chord, errors


print("=" * 60)
print("Задание 4. Комбинированный метод хорд и касательных")
print("=" * 60)

a, b = 1.0, 3.0
root, n_iter, hist_tan, hist_chord, errors = combined_method(a, b, eps=1e-6)

print(f"\nОтрезок: [{a}; {b}]")
print(f"Требуемая точность: 10⁻⁶")
print(f"Количество итераций: {n_iter}")
print(f"Приближённое значение корня: {root:.10f}")


print("\nТаблица итераций:")
print("-" * 70)
print(f"{'Итерация':^10} | {'x_n (касат.)':^20} | {'x_n (хорд)':^20} | {'Погрешность':^15}")
print("-" * 70)

for i in range(len(hist_tan)):
    x_t = hist_tan[i]
    x_c = hist_chord[i] if i < len(hist_chord) else hist_chord[-1]
    err = errors[i-1] if i > 0 else abs(b - a) / 2
    print(f"{i:^10} | {x_t:^20.10f} | {x_c:^20.10f} | {err:^15.2e}")

print("-" * 70)

print(f"\nПроверка: F({root:.10f}) = {F(root):.2e}")
```
