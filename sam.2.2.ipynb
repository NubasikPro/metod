```python
def F(x):
    return 2*x**2 - 2*x - 5

def bisection_method(a, b, eps=1e-3, max_iter=100):

    if F(a) * F(b) >= 0:
        raise ValueError("На отрезке [a, b] нет корня или корней несколько")
    
    history = []
    
    for n in range(max_iter):
        c = (a + b) / 2
        history.append(c)
        
        if abs(F(c)) < eps or (b - a) / 2 < eps:
            return c, n + 1, history
        
        if F(a) * F(c) < 0:
            b = c
        else:
            a = c
    
    return c, max_iter, history

print("=" * 60)
print("Задание 2. Метод половинного деления")
print("=" * 60)

a, b = 1.0, 3.0
root, n_iter, history = bisection_method(a, b, eps=1e-3)

print(f"\nОтрезок: [{a}; {b}]")
print(f"Требуемая точность: 10⁻³")
print(f"Количество итераций: {n_iter}")
print(f"Приближённое значение корня: {root:.6f}")

# Таблица итераций
print("\nТаблица итераций:")
print("-" * 50)
print(f"{'Итерация':^10} | {'c = (a+b)/2':^20} | {'F(c)':^15}")
print("-" * 50)

for i, c in enumerate(history):
    print(f"{i+1:^10} | {c:^20.6f} | {F(c):^15.6f}")

print("-" * 50)

# Проверка
print(f"\nПроверка: F({root:.6f}) = {F(root):.2e}")
```
