```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 4*x - 3*np.log(x) - 4

def phi(x):
    return 1 + 0.75 * np.log(x)

def simple_iteration(phi, x0, tol=1e-4, max_iter=100):
    print("-" * 80)
    print(f"{'n':>3} | {'x_n':>12} | {'x_{n+1} = φ(x_n)':>18} | {'f(x_{n+1})':>15} | {'|x_{n+1}-x_n|':>15}")
    print("-" * 80)
    
    x = x0
    for i in range(max_iter):
        x_next = phi(x)
        error = abs(x_next - x)
        print(f"{i+1:>3} | {x:>12.6f} | {x_next:>18.6f} | {f(x_next):>15.2e} | {error:>15.2e}")
        
        if error < tol:
            break
        x = x_next
    
    print("-" * 80)
    return x_next

print("=" * 80)
print("Упражнение 3. Метод простых итераций")
print("Уравнение: 4x - 3·ln x = 4 → 4x - 3·ln x - 4 = 0")
print("Итерационная функция: x = 1 + 0.75·ln x")
print("=" * 80)

root = simple_iteration(phi, 2.5, tol=1e-4)

print(f"\nНаибольший положительный корень: x = {root:.6f} → {root:.4f}")
print(f"Проверка: f(x) = {f(root):.2e}")

# График
x_plot = np.linspace(0.5, 3, 500)
y_plot = f(x_plot)

plt.figure(figsize=(10, 5))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='f(x) = 4x - 3·ln x - 4')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5)
plt.axvline(x=root, color='r', linestyle='--', label=f'Наибольший корень x = {root:.4f}')
plt.plot(root, f(root), 'ro', markersize=8)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Графическое отделение корней уравнения 4x - 3·ln x = 4')
plt.legend()
plt.show()
```
