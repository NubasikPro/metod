```python
import math
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
    return math.log10(x) + 6 - x*x  

def f2(x):
    return x * math.sin(x) - 1      


def y1(x):
    """y = lg x"""
    if x <= 0:
        return np.nan
    return math.log10(x)

def y2(x):
    """y = x² - 6"""
    return x*x - 6

# Функция для поиска интервалов и построения графика
def find_roots_and_plot(f, a, b, step, title, plot_type="single"):
    print(f"\n{title}:")
    
   
    intervals = []
    x = a
    while x < b:
        try:
            if f(x) * f(x + step) < 0:
                intervals.append((x, x + step))
                print(f"[{x:.2f}, {x + step:.2f}]")
        except:
            pass
        x += step
    
    
    x_vals = np.linspace(a, b, 500)
    
    plt.figure(figsize=(10, 5))
    
    if plot_type == "double":
       
        y1_vals = [y1(x) for x in x_vals]
        y2_vals = [y2(x) for x in x_vals]
        plt.plot(x_vals, y1_vals, 'b-', linewidth=2, label='y = lg x')
        plt.plot(x_vals, y2_vals, 'g-', linewidth=2, label='y = x² - 6')
    else:
       
        y_vals = [f(x) for x in x_vals]
        plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')
    
    plt.axhline(y=0, color='k', linestyle='--', linewidth=1)
    plt.axvline(x=0, color='k', linestyle='--', linewidth=1)
    plt.grid(True, alpha=0.3)
    
   
    for x1, x2 in intervals:
        plt.axvspan(x1, x2, alpha=0.3, color='red')
        plt.plot([(x1 + x2) / 2], [0], 'ro', markersize=8)
    
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title)
    plt.legend()
    plt.show()
    
    return intervals


find_roots_and_plot(f1, 0.1, 3, 0.1, "lg x + 6 = x²", plot_type="double")
find_roots_and_plot(f2, -10, 10, 0.5, "x·sin x - 1 = 0", plot_type="single")
```
