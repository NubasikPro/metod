```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Упражнение 1
def f1(x):
    return math.log10(x) - math.cos(x)

# Упражнение 2
def f2(x):
    return x*x - 3.2*x - 1

# Упражнение 3
def f3(x):
    if x <= 0:
        return np.nan
    return 4*x - 3*math.log(x) - 4

# Графики
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

x1 = np.linspace(0.1, 10, 500)
y1 = [f1(x) for x in x1]
axes[0].plot(x1, y1)
axes[0].axhline(y=0, color='k', ls='--')
axes[0].set_title('lg x - cos x = 0')

x2 = np.linspace(-2, 5, 500)
y2 = f2(x2)
axes[1].plot(x2, y2)
axes[1].axhline(y=0, color='k', ls='--')
axes[1].set_title('x² - 3.2x = 1')

x3 = np.linspace(0.1, 5, 500)
y3 = [f3(x) for x in x3]
axes[2].plot(x3, y3)
axes[2].axhline(y=0, color='k', ls='--')
axes[2].set_title('4x - 3ln x = 4')

for ax in axes:
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
