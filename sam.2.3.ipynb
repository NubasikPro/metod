```python
import math

def F(x):
    return 2*x**2 - 2*x - 5

def phi(x):
    """Приводим уравнение к виду x = φ(x): x = √(x + 2.5)"""
    return math.sqrt(x + 2.5)

def phi_prime(x):
    """Производная φ'(x) для проверки сходимости"""
    return 1 / (2 * math.sqrt(x + 2.5))

def simple_iteration(x0, eps=1e-6, max_iter=100):
    
    history = [x0]
    
    for n in range(max_iter):
        x1 = phi(x0)
        history.append(x1)
        
        if abs(x1 - x0) < eps:
            return x1, n + 1, history
        
        x0 = x1
    
    return x0, max_iter, history

print("=" * 60)
print("Задание 3. Метод простой итерации")
print("=" * 60)


x_test = 2.5
print(f"\nПроверка условия сходимости на отрезке [2; 3]:")
print(f"|φ'(x)| ≤ {abs(phi_prime(x_test)):.3f} < 1 — условие выполнено")


x0 = 2.0
root, n_iter, history = simple_iteration(x0, eps=1e-6)

print(f"\nНачальное приближение: x0 = {x0}")
print(f"Требуемая точность: 10⁻⁶")
print(f"Количество итераций: {n_iter}")
print(f"Приближённое значение корня: {root:.10f}")


print("\nТаблица итераций:")
print("-" * 40)
print(f"{'Итерация':^10} | {'x_n':^20}")
print("-" * 40)

for i, x_val in enumerate(history):
    print(f"{i:^10} | {x_val:^20.10f}")

print("-" * 40)


print(f"\nПроверка: F({root:.10f}) = {F(root):.2e}")
```
