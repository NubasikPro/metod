```python
import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return x**2 - 3.2*x - 1

def df(x):
    return 2*x - 3.2

# Метод Ньютона 
def newton_method(f, df, x0, tol=1e-4):
    print("-" * 60)
    print(f"{'n':>3} | {'x_n':>12} | {'f(x_n)':>15}")
    print("-" * 60)
    
    x = x0
    for i in range(20):
        fx = f(x)
        print(f"{i+1:>3} | {x:>12.6f} | {fx:>15.2e}")
        
        if abs(fx) < tol:
            break
        
        x = x - fx / df(x)
    
    print("-" * 60)
    return x

print("=" * 60)
print("Упражнение 2. x² - 3.2x = 1")
print("Метод Ньютона")
print("=" * 60)

# Положительный корень
x_pos = newton_method(f, df, 4.0, tol=1e-4)
print(f"\nПоложительный корень: x = {x_pos:.4f}")
print(f"Проверка: {x_pos:.4f}² - 3.2·{x_pos:.4f} = {x_pos**2 - 3.2*x_pos:.4f} (должно быть 1)")

print("\n" + "=" * 60)

# Отрицательный корень
x_neg = newton_method(f, df, -1.0, tol=1e-4)
print(f"\nОтрицательный корень: x = {x_neg:.4f}")
print(f"Проверка: {x_neg:.4f}² - 3.2·{x_neg:.4f} = {x_neg**2 - 3.2*x_neg:.4f} (должно быть 1)")

# График
x_plot = np.linspace(-2, 5, 500)
y_plot = f(x_plot)

plt.figure(figsize=(10, 5))
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='f(x) = x² - 3.2x - 1')
plt.axhline(y=0, color='k', linestyle='-', linewidth=0.5)
plt.axvline(x=x_pos, color='r', linestyle='--', label=f'Корень 1 = {x_pos:.4f}')
plt.axvline(x=x_neg, color='g', linestyle='--', label=f'Корень 2 = {x_neg:.4f}')
plt.plot(x_pos, 0, 'ro', markersize=8)
plt.plot(x_neg, 0, 'go', markersize=8)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Корни уравнения x² - 3.2x = 1')
plt.legend()
plt.show()
```
